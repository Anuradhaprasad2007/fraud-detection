## EDA findings
- Fraud rate: 0.129% (8,213 of 6,362,620 transactions)
- Fraud occurs exclusively in CASH_OUT and TRANSFER transaction types
- PAYMENT, CASH_IN, and DEBIT contain zero fraud cases

df.groupby('isFraud')['amount'].describe()

```python
df[df['isFraud']==1][['oldbalanceOrg','newbalanceOrig']].describe()
```

- Fraudulent transactions average ~8x higher amount than legitimate ones (₹1.47M vs ₹178K mean)
- In most fraud cases, the sender's account balance is emptied to zero after the transaction
- Per dataset documentation, balance columns should not be used as model features since fraud transactions are reversed/cancelled, making these columns leak the answer rather than reflect real pre-fraud signal
